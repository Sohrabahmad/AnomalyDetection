```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler

import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
X = df.drop("Outlier_label", axis=1)
y = df["Outlier_label"]
```

```python
X.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.ensemble import IsolationForest

clf = IsolationForest(
        n_estimators=200,
        contamination=0.036,
        random_state=42
    )
```

```python
labels = clf.fit_predict(X_scaled)
```

```python
labels
```

```python
#Visualize

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X_pca[:,0], X_pca[:,1],c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Anomaly using IsolationForest")
```

```python
import numpy as np

n_outliers = np.sum(labels == -1)
n_normal = np.sum(labels == 1)

print("outlier = ",n_outliers)
print("normal points = ",n_normal)
```

## LOF

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
neighbors = LocalOutlierFactor(contamination=0.036)

labels = neighbors.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X_pca[:,0], X_pca[:,1],c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Anomaly using LOF")
```

```python
n_outliers = np.sum(labels == -1)
n_normal = np.sum(labels == 1)

print("outlier = ",n_outliers)
print("normal points = ",n_normal)
```
